```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Logistic growth differential equation
def logistic(N, t, r, K):
    dNdt = r * N * (1 - N / K)
    return dNdt

# Parameters
r = 0.7        # growth rate
K = 150        # carrying capacity
N0 = 10        # initial population

# Time points
t = np.linspace(0, 50, 500)

# Solve ODE
solution = odeint(logistic, N0, t, args=(r, K))

# Plot solution
plt.figure(figsize=(8,5))
plt.plot(t, solution, label="Population N(t)")
plt.axhline(y=K, color='r', linestyle='--', label="Carrying Capacity (K)")
plt.xlabel("Time")
plt.ylabel("Population")
plt.title("Logistic Growth Model")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
r = 0.5
K = 100

# Logistic solution (exact formula)
def logistic_solution(t, N0, r, K):
    return (N0 * K * np.exp(r*t)) / (K + N0 * (np.exp(r*t) - 1))

# Time range
t = np.linspace(0, 20, 400)

# Different initial conditions
initial_conditions = [40,65,110,130]

# Plot
plt.figure(figsize=(8,5))

for N0 in initial_conditions:
    N_t = logistic_solution(t, N0, r, K)
    plt.plot(t, N_t, label=f"N0 = {N0}")

# Plot carrying capacity
plt.axhline(y=K, linestyle='--', label="K (carrying capacity)")

# Labels and styling
plt.xlabel("Time (t)")
plt.ylabel("Population N(t)")
plt.title("Solution of the Logistic Growth ")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Logistic model
def logistic(N, r, K):
    return r * N * (1 - N/K)

# Euler method simulation
def simulate_logistic(N0, r, K, t_max=50, dt=0.01):

    t = np.arange(0, t_max, dt)
    N = np.zeros(len(t))
    N[0] = N0

    for i in range(1, len(t)):
        N[i] = N[i-1] + dt * logistic(N[i-1], r, K)

    return t, N


# Parameters
K = 100
N0 = 10

r_values = [0.2, 0.5, 1.1]

plt.figure(figsize=(8,5))

for r in r_values:
    t, N = simulate_logistic(N0, r, K)
    plt.plot(t, N, label=f"r = {r}")

plt.axhline(K, linestyle='--', label="Carrying capacity K")

plt.xlabel("Time")
plt.ylabel("Population N(t)")
plt.title("Effect of Growth Rate r on Logistic Growth")
plt.legend()
plt.grid()

plt.show()
```
